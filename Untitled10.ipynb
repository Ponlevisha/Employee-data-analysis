```python
import pandas as pd
df=pd.read_csv("Employee Dataset - Full.csv")
df.head()
```

```python
df["Department"].value_counts()
```

```python
avg = df.groupby("Department")["Monthly Income"].mean()
print(avg)
print("Highest:", avg.idxmax(), avg.max())
print("Lowest:", avg.idxmin(), avg.min())
```

```python
top_five = df.nlargest(5, "Performance Rating")
print(top_five[["Employee Name", "Department", "Performance Rating"]])
```

```python
import matplotlib.pyplot as plt
df.plot(x="Years of Experience", y="Monthly Income", kind="scatter")
plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Monthly Income")
plt.show()
```
